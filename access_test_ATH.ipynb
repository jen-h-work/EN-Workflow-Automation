```python
# Cell 3 — Accessibility check
def check_accessibility(url: str) -> bool:
    try:
        response = requests.head(url, headers=HEADERS, timeout=10, verify=False)
        return response.status_code == 200
    except requests.exceptions.RequestException as e:
        print(f"Connection error: {e}")
        return False
```

```python
# Cell 4 — Fetch page
def fetch_page(url: str) -> str | None:
    try:
        response = requests.get(url, headers=HEADERS, timeout=10, verify=False)
        response.raise_for_status()
        return response.text
    except requests.exceptions.RequestException as e:
        print(f"Failed to fetch page: {e}")
        return None
```

```python
# Cell 5 — Main execution
if check_accessibility(URL):
    print("✅ Site is accessible")
    html = fetch_page(URL)
    if html:
        print(f"✅ Page fetched successfully ({len(html):,} characters)")
        print("\n--- First 500 characters of HTML ---")
        print(html[:500])  # Preview a snippet so you're not flooded with output
else:
    print("❌ Site is not accessible — check the URL or your network connection")
```

## Extract content from proSapient

```python
# Cell 1 — Imports
import os
os.environ['WDM_SSL_VERIFY'] = '0'

from selenium import webdriver
from selenium.webdriver.edge.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from webdriver_manager.microsoft import EdgeChromiumDriverManager
# Newer versions of webdriver-manager use EdgeChromiumDriverManager (with "Driver" in the name)
import time
```

```python
# Cell 2 — Configuration
URL = "https://platform.prosapient.com/client/projects/ef3ae048-29ac-46c6-9b36-66a2a3a9fd60/experts/c2ce1c89-6914-44ea-97fd-0f86fc484d4b?ga=0q13m5I4yLRCxbIbSdEz-Ohz2bQKsG7i7ixmKnxgVGtcjVFLLCHBZaUNkUHhA3Lm&utm_campaign=Expert_list_external_link&utm_medium=link&utm_source=bio_email"
```

```python
# Cell 3 — Browser setup
def create_driver():
    options = webdriver.EdgeOptions()
    options.add_argument("--ignore-certificate-errors")
    options.add_argument("--start-maximized")
    
    driver = webdriver.Edge(
        service=Service(EdgeChromiumDriverManager().install()),
        # Updated to match the new class name
        options=options
    )
    return driver
```

```python


# Cell 4 — Fetch full page content (with Load More clicking)
def fetch_full_content(url: str) -> str:
    driver = create_driver()
    try:
        driver.get(url)
        WebDriverWait(driver, 10).until(
            EC.presence_of_element_located((By.TAG_NAME, "body"))
        )
        time.sleep(3)

        # Keep clicking "Load more" until all experts are loaded
        while True:
            try:
                load_more = driver.find_element(By.XPATH, "//button[contains(text(), 'Load more')]")
                # Finds the "Load more" button by its text
                load_more.click()
                time.sleep(2)
                # Waits for new experts to load after each click
            except:
                break
                # Exits loop when no more "Load more" button is found

        content = driver.find_element(By.TAG_NAME, "body").text
        return content
    except Exception as e:
        print(f"Error: {e}")
        return ""
    finally:
        driver.quit()

```

```python
# Cell 5a — Parse experts from page content
def parse_experts(content: str) -> list:
    experts = []
    lines = content.split('\n')
    # Splits the full page text into individual lines

    i = 0
    while i < len(lines):
        line = lines[i].strip()

        # Detect expert name lines (start with bullet point)
        if line.startswith('•'):
            name = line.replace('•', '').strip()
            # Removes the bullet point and whitespace to get clean name

            # Role/company is the line immediately before the name
            role_company = lines[i-1].strip() if i > 0 else ""

            # Segment and geo are on lines after the name
            geo = lines[i+1].strip() if i+1 < len(lines) else ""
            segment = lines[i+2].strip() if i+2 < len(lines) else ""

            experts.append({
                "name": name,
                "role_company": role_company,
                "geography": geo,
                "segment": segment,
            })
        i += 1

    return experts
```

```python
# Cell 6 — Display results
content = fetch_full_content(URL)
experts = parse_experts(content)

print(f"Total experts found: {len(experts)}\n")
print(f"{'Name':<45} {'Geography':<20} {'Segment':<15} {'Role/Company'}")
print("-" * 130)
for e in experts:
    print(f"{e['name']:<45} {e['geography']:<20} {e['segment']:<15} {e['role_company']}")
```

```python
# Cell 7 — Check what links exist on the page
def inspect_links(url: str) -> None:
    driver = create_driver()
    try:
        driver.get(url)
        WebDriverWait(driver, 10).until(
            EC.presence_of_element_located((By.TAG_NAME, "body"))
        )
        time.sleep(3)

        # Find ALL links on the page
        all_links = driver.find_elements(By.TAG_NAME, "a")
        # Gets every hyperlink on the page

        print(f"Total links found: {len(all_links)}\n")
        for link in all_links:
            href = link.get_attribute("href")
            text = link.text.strip()
            if href:
                print(f"Text: {text:<40} URL: {href}")
                # Prints each link's visible text and its URL

    except Exception as e:
        print(f"Error: {e}")
    finally:
        driver.quit()

inspect_links(URL)
```

```python
# Cell 4 — Fixed fetch function
def fetch_all_experts(url: str) -> list:
    driver = create_driver()
    experts = []

    try:
        driver.get(url)
        WebDriverWait(driver, 10).until(
            EC.presence_of_element_located((By.TAG_NAME, "body"))
        )
        time.sleep(3)

        # Click "Load more" until all experts are visible
        while True:
            try:
                load_more = driver.find_element(By.XPATH, "//button[contains(text(), 'Load more')]")
                driver.execute_script("arguments[0].click();", load_more)
                # Uses JavaScript to click instead of Selenium — bypasses navbar blocking
                time.sleep(2)
            except:
                break

        # Get total count of bullet point elements
        total = len(driver.find_elements(By.XPATH, "//*[contains(text(), '•')]"))
        print(f"Total experts found: {total}")

        for i in range(total):
            try:
                # Re-find ALL bullet elements fresh each iteration
                # This prevents stale element errors after page re-renders
                expert_elements = driver.find_elements(By.XPATH, "//*[contains(text(), '•')]")

                if i >= len(expert_elements):
                    break

                expert_el = expert_elements[i]
                name = expert_el.text.replace('•', '').strip()
                name = ''.join(c for c in name if c.isalnum() or c in ' .,()- ')
                # Cleans special characters like ★ from names

                # Scroll element into view with offset to avoid navbar overlap
                driver.execute_script(
                    "arguments[0].scrollIntoView(true); window.scrollBy(0, -150);",
                    expert_el
                )
                # Scrolls to element then nudges up 150px to clear the fixed navbar
                time.sleep(1)

                # Use JavaScript click to bypass any overlay blocking
                driver.execute_script("arguments[0].click();", expert_el)
                time.sleep(2)

                # Capture page content after expanding profile
                content = driver.find_element(By.TAG_NAME, "body").text
                lines = content.split('\n')

                # Find expert name position in expanded content
                clean_name = name.strip()
                first_word = clean_name.split()[0] if clean_name.split() else ""
                name_idx = next(
                    (idx for idx, l in enumerate(lines)
                     if first_word in l and '•' not in l and len(l) < 80),
                    None
                )

                if name_idx is None:
                    print(f"Could not locate {name} in page content, skipping")
                    continue

                # Extract fields relative to name position
                role_company = lines[name_idx - 1].strip() if name_idx > 0 else ""
                geo          = lines[name_idx + 1].strip() if name_idx + 1 < len(lines) else ""
                segment      = lines[name_idx + 2].strip() if name_idx + 2 < len(lines) else ""
                rate         = lines[name_idx + 3].strip() if name_idx + 3 < len(lines) else ""
                status       = lines[name_idx + 4].strip() if name_idx + 4 < len(lines) else ""

                # Extract screener Q&A — lines starting with a number and period e.g. "1."
                screener = []
                in_screener = False
                for line in lines[name_idx:]:
                    if re.match(r'^\d+\.', line.strip()):
                        screener.append(line.strip())
                        in_screener = True
                    elif in_screener and line.strip() and not re.match(r'^\d+\.', line.strip()):
                        screener.append("  -> " + line.strip())
                        # Captures answer lines immediately following each question

                experts.append({
                    "name": name,
                    "role_company": role_company,
                    "geography": geo,
                    "segment": segment,
                    "rate": rate,
                    "status": status,
                    "screener_responses": screener
                })

                print(f"Extracted ({i+1}/{total}): {name}")

            except Exception as e:
                print(f"Error on expert {i+1}: {e}")
                continue

    except Exception as e:
        print(f"Fatal error: {e}")
    finally:
        driver.quit()

    return experts
```

```python
# Cell 5 — Display results
def display_experts(experts: list) -> None:
    for e in experts:
        print("=" * 80)
        print(f"Name:        {e['name']}")
        print(f"Role:        {e['role_company']}")
        print(f"Geography:   {e['geography']}")
        print(f"Segment:     {e['segment']}")
        print(f"Rate:        {e['rate']}")
        print(f"Status:      {e['status']}")
        if e['screener_responses']:
            print("Screener Responses:")
            for q in e['screener_responses']:
                print(f"  {q}")
        else:
            print("Screener Responses: None")
    print("=" * 80)
    print(f"\nTotal experts extracted: {len(experts)}")
```
